# Bird counts at restored and unrestored fynbos sites

Mean count per visit at six sites, three restored and three unrestored, each visited four times. The data are synthetic.

```python
%load_ext autoreload
%autoreload 2
```

```python
import matplotlib.pyplot as plt

from utils import load_counts, summarise_sites
```

## Load the data

One row per site visit. Visits with no count are dropped.

```python
counts = load_counts("data/bird_counts.csv")
counts.head()
```

## Mean count per site

```python
site_means = summarise_sites(counts)
site_means
```

## Plot

```python
colours = {"restored": "#2a7f62", "unrestored": "grey"}

fig, ax = plt.subplots()
ax.bar(
    site_means["site"],
    site_means["mean_count"],
    color=site_means["treatment"].map(colours),
)
ax.set_xlabel("Site")
ax.set_ylabel("Mean count per visit")
```
